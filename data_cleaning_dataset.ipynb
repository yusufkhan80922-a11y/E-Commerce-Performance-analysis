```python
datasets = {
    "customers": customers,
    "orders": orders,
    "returns": returns,
    "products": products
}
print("succesfull")
```

```python
# Convert all column names to lowercase and replace spaces with underscores

customers.columns = customers.columns.str.strip().str.lower().str.replace(" ", "_")
orders.columns = orders.columns.str.strip().str.lower().str.replace(" ", "_")
returns.columns = returns.columns.str.strip().str.lower().str.replace(" ", "_")
products.columns = products.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
customers.to_sql(
    name="customers",
    con=engine,
    if_exists="replace",
    index=False
)

orders.to_sql(
    name="orders",
    con=engine,
    if_exists="replace",
    index=False
)

returns.to_sql(
    name="returns",
    con=engine,
    if_exists="replace",
    index=False
)

products.to_sql(
    name="products",
    con=engine,
    if_exists="replace",
    index=False
)

print("All 4 datasets uploaded successfully!")
```
